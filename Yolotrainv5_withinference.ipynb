```python
import os
import shutil
import random

# Define paths
dataset_path = "/content/drive/MyDrive/RISHIK_DATA/train/"
images_path = os.path.join(dataset_path, "images")
labels_path = os.path.join(dataset_path, "labels")

# Define new directory for train/test/val split
output_dir = "/content/drive/MyDrive/RISHIKDATA_train_test_split"
os.makedirs(output_dir, exist_ok=True)

# Define new train/val/test directories
train_images_path = os.path.join(output_dir, "images/train")
val_images_path = os.path.join(output_dir, "images/val")
test_images_path = os.path.join(output_dir, "images/test")
train_labels_path = os.path.join(output_dir, "labels/train")
val_labels_path = os.path.join(output_dir, "labels/val")
test_labels_path = os.path.join(output_dir, "labels/test")

# Create train/val/test directories
os.makedirs(train_images_path, exist_ok=True)
os.makedirs(val_images_path, exist_ok=True)
os.makedirs(test_images_path, exist_ok=True)
os.makedirs(train_labels_path, exist_ok=True)
os.makedirs(val_labels_path, exist_ok=True)
os.makedirs(test_labels_path, exist_ok=True)

# Get all image files
image_files = [f for f in os.listdir(images_path) if f.endswith(('.jpg', '.png', '.jpeg'))]

# Shuffle and split
random.shuffle(image_files)
train_split = int(0.7 * len(image_files))  # 70% train
val_split = int(0.2 * len(image_files))    # 20% validation
# Remaining 10% will be test

train_files = image_files[:train_split]
val_files = image_files[train_split:train_split + val_split]
test_files = image_files[train_split + val_split:]

# Function to move files
def move_files(files, images_source, labels_source, images_dest, labels_dest):
    for file in files:
        # Move image
        shutil.copy(os.path.join(images_source, file), os.path.join(images_dest, file))
        # Move corresponding label
        label_file = file.replace(".jpg", ".txt").replace(".png", ".txt").replace(".jpeg", ".txt")
        if os.path.exists(os.path.join(labels_source, label_file)):
            shutil.copy(os.path.join(labels_source, label_file), os.path.join(labels_dest, label_file))

# Move files to train, val, and test folders
move_files(train_files, images_path, labels_path, train_images_path, train_labels_path)
move_files(val_files, images_path, labels_path, val_images_path, val_labels_path)
move_files(test_files, images_path, labels_path, test_images_path, test_labels_path)

print("Dataset successfully split into train, val, and test sets!")
```

```python
import torch
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from pathlib import Path
import random

# Load YOLOv5 model
def load_yolov5_model(weights_path):
    """
    Load the YOLOv5 model from the given weights path.

    Args:
        weights_path (str): Path to the YOLOv5 model weights (e.g., 'best.pt').

    Returns:
        model: Loaded YOLOv5 model.
    """
    try:
        model = torch.hub.load('ultralytics/yolov5', 'custom', path=weights_path, force_reload=True)
        return model
    except Exception as e:
        print(f"Error loading YOLOv5 model: {e}")
        return None

# Preprocess image
def preprocess_image(image_path):
    """
    Load and preprocess the image for YOLOv5 inference.

    Args:
        image_path (str): Path to the input image.

    Returns:
        image (numpy.ndarray): Preprocessed image.
    """
    if not os.path.exists(image_path):
        print(f"Error: Image path '{image_path}' does not exist!")
        return None

    image = cv2.imread(image_path)
    if image is None:
        print(f"Error: Failed to load image from '{image_path}'")
        return None

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convert to RGB
    return image

# Generate unique colors for each class
def generate_class_colors(class_names):
    """
    Generate a unique color for each class.

    Args:
        class_names (dict): Dictionary of class names from YOLO model.

    Returns:
        class_colors (dict): Dictionary mapping class IDs to RGB colors.
    """
    random.seed(42)  # Ensure consistent colors
    num_classes = len(class_names)
    class_colors = {}

    for class_id in range(num_classes):
        # Generate a random color
        color = tuple(random.randint(0, 255) for _ in range(3))
        class_colors[class_id] = color

    return class_colors

# Perform inference
def run_inference(model, image):
    """
    Run YOLOv5 inference on the input image.

    Args:
        model: Loaded YOLOv5 model.
        image (numpy.ndarray): Input image.

    Returns:
        results: YOLOv5 inference results.
    """
    if model is None or image is None:
        print("Error: Model or image is None. Cannot run inference.")
        return None

    results = model(image)
    df = results.pandas().xyxy[0]
    print(df)
    return results

# Draw bounding boxes on the image with different colors per class
def draw_boxes(image, results, class_colors):
    """
    Draw bounding boxes and labels on the image with different colors for each class.

    Args:
        image (numpy.ndarray): Input image.
        results: YOLOv5 inference results.
        class_colors (dict): Dictionary mapping class IDs to RGB colors.

    Returns:
        image (numpy.ndarray): Image with bounding boxes and labels.
    """
    if results is None:
        print("Error: No results found for bounding boxes.")
        return image

    for *box, conf, cls in results.xyxy[0]:
        class_id = int(cls)
        label = f"{results.names[class_id]} {conf:.2f}"
        x1, y1, x2, y2 = map(int, box)

        # Get the color for this class
        color = class_colors.get(class_id, (255, 0, 0))  # Default to blue if class not found

        # Draw bounding box
        cv2.rectangle(image, (x1, y1), (x2, y2), color, 2)
        cv2.putText(image, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, color, 2)

    return image

# Save and display the results
def save_and_show_results(image, output_path):
    """
    Save and display the results.

    Args:
        image (numpy.ndarray): Image with bounding boxes and labels.
        output_path (str): Path to save the output image.
    """
    # Ensure the output directory exists
    output_dir = os.path.dirname(output_path)
    os.makedirs(output_dir, exist_ok=True)

    # Ensure the file extension is valid
    if not output_path.lower().endswith(('.jpg', '.png')):
        print("Error: Output file must have a valid image extension (.jpg or .png).")
        return

    # Save the image
    try:
        success = cv2.imwrite(output_path, cv2.cvtColor(image, cv2.COLOR_RGB2BGR))
        if not success:
            print(f"Error: Failed to save image at '{output_path}'")
            return
        print(f"Image saved successfully at: {output_path}")

        # Show the image
        plt.imshow(image)
        plt.axis('off')
        plt.show()
    except Exception as e:
        print(f"Error saving or displaying the image: {e}")

# Main function
def detect_objects(image_path, weights_path, output_path):
    """
    Main function to detect objects in an image using YOLOv5.

    Args:
        image_path (str): Path to the input image.
        weights_path (str): Path to the YOLOv5 model weights.
        output_path (str): Path to save the output image.
    """
    # Load model
    model = load_yolov5_model(weights_path)
    if model is None:
        return

    # Generate colors for each class
    class_colors = generate_class_colors(model.names)

    # Preprocess image
    image = preprocess_image(image_path)
    if image is None:
        return

    # Run inference
    results = run_inference(model, image)
    if results is None:
        return

    # Draw bounding boxes
    image_with_boxes = draw_boxes(image, results, class_colors)

    # Save and display results
    save_and_show_results(image_with_boxes, output_path)

# Example usage
if __name__ == "__main__":
    # Paths
    weights_path = "/content/drive/MyDrive/trainyolo/exp2/weights/best.pt"  # Path to trained weights
    image_path = "/content/drive/MyDrive/Sachidananda(KNCVB10072)_Annotationdata/Final_imagedatannotations/images/KNCVB10072_001.jpg"  # Path to input image
    output_path = "/content/drive/MyDrive/Output/output_detected.jpg"  # Save with .jpg extension

    # Run detection
    detect_objects(image_path, weights_path, output_path)
```

```python
import torch
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# Define 17 distinct colors for bounding boxes
CLASS_COLORS = {
    0: (255, 0, 0),      # Red
    1: (0, 255, 0),      # Green
    2: (0, 0, 255),      # Blue
    3: (255, 255, 0),    # Yellow
    4: (255, 0, 255),    # Magenta
    5: (0, 255, 255),    # Cyan
    6: (128, 0, 0),      # Dark Red
    7: (0, 128, 0),      # Dark Green
    8: (0, 0, 128),      # Dark Blue
    9: (128, 128, 0),    # Olive
    10: (128, 0, 128),   # Purple
    11: (0, 128, 128),   # Teal
    12: (192, 192, 192), # Silver
    13: (128, 128, 128), # Gray
    14: (255, 165, 0),   # Orange
    15: (75, 0, 130),    # Indigo
    16: (0, 191, 255)    # Deep Sky Blue
}

# Load YOLOv5 model
def load_yolov5_model(weights_path):
    """
    Load the YOLOv5 model from the given weights path.
    """
    try:
        model = torch.hub.load('ultralytics/yolov5', 'custom', path=weights_path, force_reload=True)
        return model
    except Exception as e:
        print(f"Error loading YOLOv5 model: {e}")
        return None

# Preprocess image
def preprocess_image(image_path):
    """
    Load and preprocess the image for YOLOv5 inference.
    """
    if not os.path.exists(image_path):
        print(f"Error: Image path '{image_path}' does not exist!")
        return None

    image = cv2.imread(image_path)
    if image is None:
        print(f"Error: Failed to load image from '{image_path}'")
        return None

    return cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convert to RGB

# Run inference
def run_inference(model, image):
    """
    Run YOLOv5 inference on the input image.
    """
    if model is None or image is None:
        print("Error: Model or image is None. Cannot run inference.")
        return None

    return model(image)

# Draw bounding boxes with distinct colors
def draw_boxes(image, results):
    """
    Draw bounding boxes and labels on the image with different colors for each class.
    """
    if results is None:
        print("Error: No results found for bounding boxes.")
        return image

    for *box,conf, cls in results.xyxy[0]:
        class_id = int(cls)
        label = f"{results.names[class_id]} {conf:.2f}"
        x1, y1, x2, y2 = map(int, box)

        # Get color for this class (default to white if class_id is out of range)
        color = CLASS_COLORS.get(class_id, (255, 255, 255))

        # Draw bounding box
        cv2.rectangle(image, (x1, y1), (x2, y2), color, 2)
        cv2.putText(image, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, color, 2)

    return image

# Save and display the results
def save_and_show_results(image, output_path):
    """
    Save and display the results.
    """
    os.makedirs(os.path.dirname(output_path), exist_ok=True)

    try:
        cv2.imwrite(output_path, cv2.cvtColor(image, cv2.COLOR_RGB2BGR))
        print(f"Image saved successfully at: {output_path}")

        # Show image
        plt.imshow(image)
        plt.axis('off')
        plt.show()
    except Exception as e:
        print(f"Error saving or displaying the image: {e}")

# Main function
def detect_objects(image_path, weights_path, output_path):
    """
    Detect objects in an image using YOLOv5.
    """
    model = load_yolov5_model(weights_path)
    if model is None:
        return

    image = preprocess_image(image_path)
    if image is None:
        return

    results = run_inference(model, image)
    if results is None:
        return

    image_with_boxes = draw_boxes(image, results)
    save_and_show_results(image_with_boxes, output_path)

# Example usage
if __name__ == "__main__":
    weights_path = "/content/drive/MyDrive/trainyolo/exp2/weights/best.pt"
    image_path = "/content/drive/MyDrive/Sachidananda(KNCVB10072)_Annotationdata/Final_imagedatannotations/images/KNCVB10072_189.jpg"
    output_path = "/content/drive/MyDrive/Output/output_detected.jpg"

    detect_objects(image_path, weights_path, output_path)
```

```python
import torch
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from tqdm import tqdm  # For progress bar

# Load YOLOv5 model
def load_yolov5_model(weights_path):
    """
    Load the YOLOv5 model from the given weights path.

    Args:
        weights_path (str): Path to the YOLOv5 model weights (e.g., 'best.pt' or 'last.pt').

    Returns:
        model: Loaded YOLOv5 model.
    """
    model = torch.hub.load('ultralytics/yolov5', 'custom', path=weights_path)
    return model

# Preprocess image
def preprocess_image(image_path):
    """
    Load and preprocess the image for YOLOv5 inference.

    Args:
        image_path (str): Path to the input image.

    Returns:
        image (numpy.ndarray): Preprocessed image.
    """
    image = cv2.imread(image_path)
    if image is None:
        print(f"Error: Unable to load image at {image_path}. Skipping...")
        return None
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convert to RGB
    return image

# Perform inference
def run_inference(model, image):
    """
    Run YOLOv5 inference on the input image.

    Args:
        model: Loaded YOLOv5 model.
        image (numpy.ndarray): Input image.

    Returns:
        results: YOLOv5 inference results.
    """
    results = model(image)
    return results

# Draw bounding boxes on the image
def draw_boxes(image, results):
    """
    Draw bounding boxes and labels on the image.

    Args:
        image (numpy.ndarray): Input image.
        results: YOLOv5 inference results.

    Returns:
        image (numpy.ndarray): Image with bounding boxes and labels.
    """
    for *box, conf, cls in results.xyxy[0]:
        label = f"{results.names[int(cls)]} {conf:.2f}"
        x1, y1, x2, y2 = map(int, box)
        cv2.rectangle(image, (x1, y1), (x2, y2), (255, 0, 0), 2)  # Draw bounding box
        cv2.putText(image, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (255, 0, 0), 2)  # Add label
    return image

# Save and display the results
def save_and_show_results(image, output_path):
    """
    Save and display the results.

    Args:
        image (numpy.ndarray): Image with bounding boxes and labels.
        output_path (str): Path to save the output image.
    """
    cv2.imwrite(output_path, cv2.cvtColor(image, cv2.COLOR_RGB2BGR))  # Save image
    plt.imshow(image)
    plt.axis('off')
    plt.show()

# Main function to process multiple images
def detect_objects_in_directory(image_dir, weights_path, output_dir):
    """
    Detect objects in all images in the specified directory.

    Args:
        image_dir (str): Path to the directory containing input images.
        weights_path (str): Path to the YOLOv5 model weights.
        output_dir (str): Path to save the output images.
    """
    # Load model
    model = load_yolov5_model(weights_path)

    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)

    # Get a list of all image paths in the directory
    image_paths = [os.path.join(image_dir, fname) for fname in os.listdir(image_dir) if fname.endswith(('.jpg', '.png', '.jpeg'))]

    # Process each image
    for image_path in tqdm(image_paths, desc="Processing images"):
        # Preprocess image
        image = preprocess_image(image_path)
        if image is None:
            continue  # Skip if the image couldn't be loaded

        # Run inference
        results = run_inference(model, image)

        # Draw bounding boxes
        image_with_boxes = draw_boxes(image, results)

        # Save results
        output_path = os.path.join(output_dir, os.path.basename(image_path))
        save_and_show_results(image_with_boxes, output_path)

# Example usage
if __name__ == "__main__":
    # Paths
    weights_path = "/content/drive/MyDrive/trainyolo/exp2/weights/best.pt"  # Path to your trained weights
    image_dir = "/content/drive/MyDrive/Sachidananda(KNCVB10072)_Annotationdata/Final_imagedatannotations/images"  # Directory containing input images
    output_dir = "/content/drive/MyDrive/Sachidananda(KNCVB10072)_Annotationdata/Final_imagedatannotations/output"  # Directory to save output images

    # Run detection on all images in the directory
    detect_objects_in_directory(image_dir, weights_path, output_dir)
```
